# 1️⃣5️⃣ Feature Engineering

## Purpose

Feature Engineering is the process of creating new meaningful features from the existing dataset to improve the predictive performance of machine learning models.

In this project, additional intelligent features are created to provide investment analysis, property quality assessment, and recommendation capabilities.

### New Features Created

- House Age
- House Size Category
- Family Suitability Score
- Property Score
- Neighborhood Average Price
- Investment Score
- Value for Money Score

```python
# ==========================================================
# House Age
# ==========================================================

CURRENT_YEAR = 2026

df["HouseAge"] = CURRENT_YEAR - df["YearBuilt"]

df[["YearBuilt", "HouseAge"]].head()
```

```python
# ==========================================================
# House Size Category
# ==========================================================

df["HouseSizeCategory"] = pd.cut(
    df["SquareFeet"],
    bins=[0, 1000, 2000, 3500, float("inf")],
    labels=["Small", "Medium", "Large", "Luxury"]
)

df["HouseSizeCategory"].value_counts()
```

```python
# ==========================================================
# Family Suitability Score
# ==========================================================

df["FamilySuitabilityScore"] = (
    df["Bedrooms"] * 3 +
    df["Bathrooms"] * 2
)

df[["Bedrooms","Bathrooms","FamilySuitabilityScore"]].head()
```

```python
# ==========================================================
# Property Score
# ==========================================================

df["PropertyScore"] = (
    df["SquareFeet"]/100 +
    df["Bedrooms"]*10 +
    df["Bathrooms"]*8 -
    df["HouseAge"]*0.5
)

df[["PropertyScore"]].head()
```

```python
# ==========================================================
# Neighborhood Average Price
# ==========================================================

df["NeighborhoodAveragePrice"] = (
    df.groupby("Neighborhood")["Price"]
      .transform("mean")
)

df[["Neighborhood","NeighborhoodAveragePrice"]].head()
```

```python
# ==========================================================
# Investment Score
# ==========================================================

df["InvestmentScore"] = (
    df["NeighborhoodAveragePrice"] /
    df["Price"]
)

df["InvestmentScore"] = df["InvestmentScore"].round(2)

df[["InvestmentScore"]].head()
```

```python
# ==========================================================
# Value For Money Score
# ==========================================================

df["ValueForMoneyScore"] = (
    df["SquareFeet"] /
    df["Price"]
) * 100000

df[["ValueForMoneyScore"]].head()
```

```python
# ==========================================================
# Newly Engineered Features
# ==========================================================

new_features = [
    "HouseAge",
    "HouseSizeCategory",
    "FamilySuitabilityScore",
    "PropertyScore",
    "NeighborhoodAveragePrice",
    "InvestmentScore",
    "ValueForMoneyScore"
]

df[new_features].head()
```

```python
# ==========================================================
# Updated Dataset Information
# ==========================================================

print("Updated Dataset Shape:", df.shape)

df.head()
```

- House Size Category classifies properties based on their area.
- Family Suitability Score estimates how suitable a house is for larger families.
- Property Score combines multiple attributes into a single quality metric.
- Neighborhood Average Price captures local market trends.

# 1️⃣7️⃣ Train-Test Split

## Purpose

Machine Learning models should not be trained and tested using the same dataset. To evaluate the model's performance on unseen data, the dataset is divided into training and testing datasets.

The training dataset is used to train the machine learning models, while the testing dataset is used to evaluate the model's prediction accuracy.

---

## Objectives

- Split the dataset into Training (70%) and Testing (30%).
- Create Feature Matrix (X) and Target Variable (y).
- Verify the dimensions of the split datasets.
- Check for missing values after splitting.
- Prepare the dataset for Machine Learning model training.

---

## Why Train-Test Split?

Train-Test Split helps us evaluate how well the machine learning model performs on unseen data and prevents overfitting.

**Training Dataset : 70%**

**Testing Dataset : 30%**

```python
# ==========================================================
# SECTION 22 - EXTRA TREES REGRESSION
# AI SMART PROPERTY ADVISOR
# ==========================================================

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("="*80)
print("SECTION 22 : EXTRA TREES REGRESSION")
print("="*80)

# ----------------------------------------------------------
# Create Extra Trees Model
# ----------------------------------------------------------

et_model = ExtraTreesRegressor(
    n_estimators=100,
    random_state=42
)

print("\nTraining Extra Trees Model...")

et_model.fit(X_train, y_train)

print("✓ Model Trained Successfully")

# ----------------------------------------------------------
# Prediction
# ----------------------------------------------------------

print("\nPredicting House Prices...")

y_pred_et = et_model.predict(X_test)

print("✓ Prediction Completed")

# ----------------------------------------------------------
# Model Evaluation
# ----------------------------------------------------------

mae_et = mean_absolute_error(y_test, y_pred_et)

mse_et = mean_squared_error(y_test, y_pred_et)

rmse_et = np.sqrt(mse_et)

r2_et = r2_score(y_test, y_pred_et)

print("\n"+"="*80)
print("MODEL PERFORMANCE")
print("="*80)

print(f"Mean Absolute Error     : {mae_et:.2f}")
print(f"Mean Squared Error      : {mse_et:.2f}")
print(f"Root Mean Squared Error : {rmse_et:.2f}")
print(f"R² Score                : {r2_et:.4f}")

# ----------------------------------------------------------
# Actual vs Predicted
# ----------------------------------------------------------

comparison_et = pd.DataFrame({

    "Actual Price": y_test.values,

    "Predicted Price": y_pred_et,

    "Difference": y_test.values - y_pred_et

})

print("\n"+"="*80)
print("ACTUAL VS PREDICTED HOUSE PRICES")
print("="*80)

print(comparison_et.head(10))

# ----------------------------------------------------------
# Scatter Plot
# ----------------------------------------------------------

plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred_et)

plt.xlabel("Actual Price")

plt.ylabel("Predicted Price")

plt.title("Extra Trees: Actual vs Predicted")

plt.grid(True)

plt.show()

# ----------------------------------------------------------
# Residual Plot
# ----------------------------------------------------------

residuals_et = y_test - y_pred_et

plt.figure(figsize=(8,6))

plt.scatter(y_pred_et, residuals_et)

plt.axhline(y=0,color='red',linestyle='--')

plt.xlabel("Predicted Price")

plt.ylabel("Residual Error")

plt.title("Extra Trees Residual Plot")

plt.grid(True)

plt.show()

# ----------------------------------------------------------
# Error Distribution
# ----------------------------------------------------------

plt.figure(figsize=(8,6))

plt.hist(residuals_et,bins=30)

plt.title("Extra Trees Error Distribution")

plt.xlabel("Prediction Error")

plt.ylabel("Frequency")

plt.grid(True)

plt.show()

# ----------------------------------------------------------
# Performance Summary
# ----------------------------------------------------------

performance_et = pd.DataFrame({

    "Metric":[
        "Mean Absolute Error",
        "Mean Squared Error",
        "Root Mean Squared Error",
        "R² Score"
    ],

    "Value":[
        mae_et,
        mse_et,
        rmse_et,
        r2_et
    ]

})

print("\n"+"="*80)
print("EXTRA TREES PERFORMANCE SUMMARY")
print("="*80)

print(performance_et)

# ----------------------------------------------------------
# Final Status
# ----------------------------------------------------------

print("\n"+"="*80)
print("EXTRA TREES MODEL COMPLETED")
print("="*80)

print("✓ Model Training Completed")
print("✓ Prediction Completed")
print("✓ Evaluation Completed")
print("✓ Visualization Completed")

print("\nNext Section : 23 - Model Comparison")

print("="*80)
```
